```python
import string   # Character sets: ascii_letters, digits, punctuation
import secrets  # Cryptographically secure randomness — enterprise standard
import math     # log2() function — Entropy calculate karne ke liye

LETTERS  = string.ascii_letters  # 52 characters — uppercase + lowercase
DIGITS   = string.digits         # 10 characters — 0 to 9
SYMBOLS  = string.punctuation    # 32 characters — special symbols

def calculate_entropy(length, pool_size):
    # E = L x log2(R) — Shannon Entropy formula
    # math.log2() = base-2 logarithm
    entropy = length * math.log2(pool_size)
    return round(entropy, 2)  # 2 decimal places pe round karo

def get_strength_label(entropy):
    # Entropy value ke hisaab se strength label do
    if entropy < 28:
        return "🔴 Very Weak   — Seconds mein crack hoga!"
    elif entropy < 36:
        return "🟠 Weak        — Minutes mein crack hoga!"
    elif entropy < 60:
        return "🟡 Reasonable  — Kuch din lagte hain"
    elif entropy < 128:
        return "🟢 Strong      — Kai saal lagte hain"
    else:
        return "💎 Very Strong — Millions of years!"


# STEP 3: PASSWORD GENERATOR — Core Engine (Slide 13 & 15)


def generate_password(length, use_letters=True, use_digits=True, use_symbols=True):
    # ----- Character Pool Build karo -----
    # Sirf woh characters include karo jo user ne choose kiye hain
    character_pool = ""  # Khali string se shuru

    if use_letters:
        character_pool += LETTERS  # 52 letters add karo
    if use_digits:
        character_pool += DIGITS   # 10 digits add karo
    if use_symbols:
        character_pool += SYMBOLS  # 32 symbols add karo

    # Pool khali hai? Matlab user ne koi option select nahi kiya
    if not character_pool:
        return None, 0, 0  # Error signal — None return karo

    pool_size = len(character_pool)  # R = character pool ka size


    # GUARANTEE: Har selected type se kam az kam 1 character zaroor ho
    # Yeh "mandatory complexity" ensure karta hai.
    # Pehle ek guaranteed character har type se lo, phir baaki randomly fill karo.
    

    guaranteed_chars = []  # Guaranteed characters ki list

    if use_letters:
        guaranteed_chars.append(secrets.choice(LETTERS))  # 1 letter guaranteed
    if use_digits:
        guaranteed_chars.append(secrets.choice(DIGITS))   # 1 digit guaranteed
    if use_symbols:
        guaranteed_chars.append(secrets.choice(SYMBOLS))  # 1 symbol guaranteed

    # Baaki remaining length ke liye random characters pool se lo
    remaining_length = length - len(guaranteed_chars)

    # List comprehension — enterprise way to build character list
    # secrets.choice() = cryptographically secure selection
    random_chars = [secrets.choice(character_pool) for _ in range(remaining_length)]

    # Saare characters ko milao — guaranteed + random
    all_chars = guaranteed_chars + random_chars

    secrets.SystemRandom().shuffle(all_chars)

    password = ''.join(all_chars)  # List ko string mein convert karo — O(N)

    # Entropy calculate karo
    entropy = calculate_entropy(length, pool_size)

    return password, entropy, pool_size

# STEP 4: INPUT VALIDATION — Phase 1 (Slide 8)


def get_valid_length():
    while True:
        try:
            length = int(input("  🔢 Password length (8-64): ").strip())

            # NIST 2024 minimum check
            if length < 8:
                print("  ⚠️  Minimum 8 characters chahiye! (NIST standard)")
                print("  💡 High security ke liye 15+ recommend kiya jaata hai.\n")
                continue

            # Maximum length check
            if length > 64:
                print("  ⚠️  Maximum 64 characters allowed hain! (NIST standard)\n")
                continue

            # NIST 2024 recommendation warning
            if length < 15:
                print(f"  ⚠️  {length} characters = weak. NIST 2024 recommends 15+")
                print("  💡 Tip: 16+ characters = millions of years to crack!\n")

            return length  # Valid length return karo

        except ValueError:
            print("  ❌ Sirf number dalo!\n")

# STEP 5: MAIN FUNCTION — Complete IPO Engine
# -----------------------------------------------------------------------------

def main():
    print("\n" + "=" * 60)
    print("  🔐 DECODELABS PASSWORD GENERATOR")
    print("  Batch 2026 | Python Project 3")
    print("  Enterprise-grade | NIST 2024 Compliant")
    print("=" * 60)

    while True:
        print("\n  ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
        print("  Kya karna chahte ho?")
        print("  1. 🔑 Password generate karo")
        print("  2. 📊 Entropy calculator (manual check)")
        print("  3. 🚪 Exit")
        print("  ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")

        choice = input("  Option chuno (1-3): ").strip()

        # =====================================================================
        if choice == "1":
        # =====================================================================
            print("\n  📐 STEP 1: Password length batao")
            length = get_valid_length()

            # Character type options
            print("\n  🎛️  STEP 2: Character types choose karo")
            print("  (Enter dabao ya 'n' likho skip karne ke liye)\n")

            # Letters
            letters_input = input("  Include letters? (A-Z, a-z) [Y/n]: ").strip().lower()
            use_letters = letters_input != 'n'  # Default: Yes

            # Digits
            digits_input = input("  Include digits? (0-9) [Y/n]: ").strip().lower()
            use_digits = digits_input != 'n'  # Default: Yes

            # Symbols
            symbols_input = input("  Include symbols? (!@#$...) [Y/n]: ").strip().lower()
            use_symbols = symbols_input != 'n'  # Default: Yes

            # Validate: kam az kam ek type honi chahiye
            if not use_letters and not use_digits and not use_symbols:
                print("\n  ❌ Kam az kam ek character type select karo!\n")
                continue

            # Kitne passwords chahiye?
            try:
                count_input = input("  Kitne passwords generate karein? [1]: ").strip()
                count = int(count_input) if count_input else 1
                count = max(1, min(count, 10))  # 1 se 10 ke beech
            except ValueError:
                count = 1  # Default: 1

      
            # PASSWORD GENERATION — secrets module ka kaam
         
            print("\n" + "=" * 60)
            print(f"  🔑 GENERATED PASSWORD(S)")
            print("=" * 60)

            for i in range(count):
                password, entropy, pool_size = generate_password(
                    length, use_letters, use_digits, use_symbols
                )

                if password is None:
                    print("  ❌ Password generate nahi ho saka!")
                    break

                strength = get_strength_label(entropy)

                # Output display — Phase 3
                if count > 1:
                    print(f"\n  [{i+1}] ", end="")
                else:
                    print(f"\n  ", end="")

                print(f"Password : {password}")
                print(f"     Length    : {length} characters")
                print(f"     Pool Size : {pool_size} characters (R)")
                print(f"     Entropy   : {entropy} bits (E = L × log₂R)")
                print(f"     Strength  : {strength}")

            print("\n" + "=" * 60)


        elif choice == "2":
 
            # Manual entropy calculator
            print("\n  📊 ENTROPY CALCULATOR")
            print("  Formula: E = L × log₂(R)\n")
            try:
                L = int(input("  Password length (L): "))
                R = int(input("  Character pool size (R): "))
                if L > 0 and R > 1:
                    E = calculate_entropy(L, R)
                    strength = get_strength_label(E)
                    print(f"\n  Entropy = {L} × log₂({R}) = {E} bits")
                    print(f"  Strength: {strength}\n")
                else:
                    print("  ⚠️  Valid values dalo!\n")
            except ValueError:
                print("  ❌ Sirf numbers dalo!\n")

       
        elif choice == "3":
      
            print("\n  👋 Password Generator band ho raha hai...")
            print("  DecodeLabs | Keep Coding! 💪\n")
            break

        else:
            print("\n  ⚠️  1, 2 ya 3 dalo!\n")


if __name__ == "__main__":
    main()

```
