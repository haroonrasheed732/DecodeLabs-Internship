```python
my_tasks = []
def generate_id():
    return len(my_tasks) + 1

def add_task(task_name):
    new_task = {
        "id"  : generate_id(), 
        "task": task_name,     
        "done": False           
    }
    my_tasks.append(new_task)
    print(f"\n  ✅ Task added: '{task_name}' (ID: {new_task['id']})")
def view_tasks():
    if len(my_tasks) == 0:
        print("\n  📭 Koi task nahi hai! Pehle kuch add karo.\n")
        return  
    print("\n" + "=" * 50)
    print("  📋 AAPKI TO-DO LIST")
    print("=" * 50)
    for index, task in enumerate(my_tasks, start=1):
        if task["done"]:
            status = "✅ Done   "   
        else:
            status = "⏳ Pending"  
        print(f"  {index}. [{status}] ID:{task['id']} — {task['task']}")
    print("=" * 50)
    print(f"  📊 Total Tasks: {len(my_tasks)}\n")

def complete_task(task_id):
    for task in my_tasks:
        if task["id"] == task_id:   
            task["done"] = True     
            print(f"\n  🎉 Task complete: '{task['task']}'")
            return  
    print(f"\n  ❌ ID {task_id} wali koi task nahi mili.")
def delete_task(task_id):
  
    global my_tasks
    original_length = len(my_tasks)
    my_tasks = [task for task in my_tasks if task["id"] != task_id]
    if len(my_tasks) < original_length:
        print(f"\n  🗑️  Task ID {task_id} delete ho gayi.")
    else:
        print(f"\n  ❌ ID {task_id} wali koi task nahi mili.")

def main():
    print("\n" + "=" * 50)
    print("  🚀 DECODELABS TO-DO LIST MANAGER")
    print("  Batch 2026 | Python Project 1")
    print("=" * 50)
    while True:
        print("\n  Kya karna chahte ho?")
        print("  1. ➕ Task add karo")
        print("  2. 📋 Saari tasks dekho")
        print("  3. ✅ Task complete karo")
        print("  4. 🗑️  Task delete karo")
        print("  5. 🚪 Exit (Program band karo)")
        print("-" * 50)
        choice = input("  Apna option chuno (1-5): ").strip()
        if choice == "1":
            task_name = input("  ✏️  Task likho: ").strip()

            if task_name: 
                add_task(task_name)
            else:
                print("\n  ⚠️  Task naam khali nahi ho sakta!")

        elif choice == "2":
            view_tasks()

        elif choice == "3":
            view_tasks()  
            try:
                task_id = int(input("  Konsi task complete hui? ID dalo: "))
                complete_task(task_id)
            except ValueError:
                print("\n  ⚠️  Sirf number dalo please!")

        elif choice == "4":
            view_tasks()  
            try:
                task_id = int(input("  Konsi task delete karni hai? ID dalo: "))
                delete_task(task_id)
            except ValueError:
                print("\n  ⚠️  Sirf number dalo please!")

        elif choice == "5":
            print("\n  👋 Program band ho raha hai...")
            print("  ⚠️  Note: RAM volatile hai — data save nahi hoga!")
            print("  DecodeLabs | Keep Coding! 💪\n")
            break  

        else:
            print("\n  ⚠️  Galat choice! 1 se 5 ke beech kuch dalo.")
if __name__ == "__main__":
    main()  # Program shuru karo!

```
