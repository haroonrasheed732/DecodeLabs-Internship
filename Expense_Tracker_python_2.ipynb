```python
# -----------------------------------------------------------------------------
# STEP 1: STATE INITIALIZATION — "Anatomy of State" (Slide 10)
# -----------------------------------------------------------------------------

expenses        = []   
total           = 0.0  
transaction_count = 0 
```

```python
# -----------------------------------------------------------------------------
# STEP 2: DISPLAY FUNCTION — Output Phase (IPO Model ka "O")
# -----------------------------------------------------------------------------
def display_summary():
    if len(expenses) == 0:
        print("\n  📭 Abhi tak koi expense add nahi hua.\n")
        return

    print("\n" + "=" * 55)
    print("  📊 EXPENSE SUMMARY")
    print("=" * 55)

    for index, exp in enumerate(expenses, start=1):
        print(f"  {index:>2}. {exp['category']:<15} Rs. {exp['amount']:>8.2f}")

    print("-" * 55)
    print(f"  {'TOTAL SPENT':<20}       Rs. {total:>8.2f}")
    print(f"  {'Transactions':<20}       {transaction_count:>8}")

    if transaction_count > 0:
        average = total / transaction_count  
        print(f"  {'Average per expense':<20}       Rs. {average:>8.2f}")

    print("=" * 55 + "\n")
```

```python
# -----------------------------------------------------------------------------
# STEP 3: ADD EXPENSE FUNCTION — Process Phase (IPO Model ka "P")
# -----------------------------------------------------------------------------


def add_expense(amount, category):
    global total, transaction_count
    expense_record = {
        "id"      : transaction_count + 1,  
        "amount"  : amount,                  
        "category": category                 
    }

    expenses.append(expense_record)

    total             += amount   
    transaction_count += 1        

    print(f"\n  ✅ Added: {category} — Rs. {amount:.2f}")
    print(f"  💰 Running Total: Rs. {total:.2f}")  
```

```python
# -----------------------------------------------------------------------------
# STEP 4: MAIN FUNCTION — Complete IPO Engine
# -----------------------------------------------------------------------------


def main():

    print("\n" + "=" * 55)
    print("  💼 DECODELABS EXPENSE TRACKER")
    print("  Batch 2026 | Python Project 2")
    print("=" * 55)
    print("  Apne expenses enter karo.")
    print("  'quit' likho jab band karna ho.\n")

    while True:

        user_input = input("  💸 Amount (ya 'quit'): ").strip().lower()

        if user_input == "quit":

            print("\n" + "=" * 55)
            print("  🔴 SESSION BAND HO RAHI HAI...")
            print("=" * 55)
            display_summary()  
            print(f"   FINAL TOTAL SPENT: Rs. {total:.2f}")
            print(f"    Note: RAM volatile hai — data ab erase ho jayega!")
            print("  DecodeLabs | Keep Coding! \n")
            break  

        try:
            amount = float(user_input)  
            if amount <= 0:
                print("    Amount positive hona chahiye! Dobara try karo.")
                continue  
            category = input("   Category (e.g. Food, Travel): ").strip()
            if not category:
                category = "General"  
            add_expense(amount, category)

        except ValueError:
            print("   Invalid amount! Sirf number dalo (e.g. 100, 50.5)")
            print("   Tip: 'quit' likhein band karne ke liye.\n")

        print()

if __name__ == "__main__":
    main()  # Program shuru karo!

```
